```python
# Leetcode: Two Sum (1.)

def twoSum(nums, target):
    """
    :type nums: List[int]
    :type target: int
    :rtype: List[int]
    """
    mem = {}
    for i in range(len(nums)):
        if target - nums[i] in mem:
            return [mem[target - nums[i]], i]
        if nums[i] not in mem:
            mem[nums[i]] = i
    
print(twoSum([1,2,3,4,5], 6))
```

## Einops (einsum, rearrange, reduce)

```python
import torch
from einops import einsum, rearrange, reduce

x = torch.rand(2, 3)
print(x)
y = 1/3 * torch.tensor([1,2,3])

print(einsum(y, x, 'i, j i -> j'))
print('---')
print(reduce(x, 'i j -> j', 'mean'))
print('---')
print(einsum(x, y, 'i j, k -> k i j'))
print('---')
print(rearrange(einsum(x, y, 'i j, k -> k i j'), 'i j k -> j (i k)'))
```
